```python
# 1. Area Chart
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

avg_petal_length = data.groupby("Species")["PetalLengthCm"].mean()

species = avg_petal_length.index
values = avg_petal_length.values

plt.stackplot(species,values,labels=species)

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")


plt.grid(True,which="major")
plt.minorticks_on()
plt.grid(True, which="minor", linestyle="--", alpha=0.5)

plt.legend(
    title="Species",
    loc = "lower right",
    fontsize = 12
)

plt.show()
```

```python
# 2. Bubble Chart
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

species_list = data["Species"].unique()

for species in species_list:
  subset = data[data["Species"] == species]

  plt.scatter(
      subset['SepalLengthCm'],
      subset['SepalWidthCm'],
      s = subset['PetalLengthCm'] *30,
      alpha = 0.4,
      label = species
  )

plt.title("Iris Flower Features")

plt.legend(
    title = "Species"
)

plt.plot()
```

```python
# Heat Map
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

numeric_data = data.select_dtypes(["number"])

corr_matrix = numeric_data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,        # Show correlation values
    cmap="coolwarm",   # Color map for better contrast
    linewidths=0.5
)

plt.title("Heatmap of Iris Features")

plt.show()
```
