### 1. Area Chart

- Create an area chart showing the average petal length for each species.

- Title: "Average Petal Length by Species"

- X-axis: Species

- Y-axis: Average Petal Length (cm)

- Add legend and gridlines (major and minor).

- Set the legend position to the bottom-right corner.

- Add a legend title "Species" and set the legend font size to 12.

```python
# 1. Area Chart
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

avg_petal_length = data.groupby("Species")["PetalLengthCm"].mean()

species = avg_petal_length.index
values = avg_petal_length.values

plt.stackplot(species,values,labels=species)

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")


plt.grid(True,which="major")
plt.minorticks_on()
plt.grid(True, which="minor", linestyle="--", alpha=0.5)

plt.legend(
    title="Species",
    loc = "lower right",
    fontsize = 12
)
plt.savefig("Area_chart.png", dpi=300, bbox_inches='tight')

plt.show()
```

### 2. Bubble Chart

- X-axis: Sepal Length

- Y-axis: Sepal Width

- Bubble size: Petal Length

- Bubble color: Species

- Title: "Iris Flower Features"


```python
# 2. Bubble Chart
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

species_list = data["Species"].unique()

for species in species_list:
  subset = data[data["Species"] == species]

  plt.scatter(
      subset['SepalLengthCm'],
      subset['SepalWidthCm'],
      s = subset['PetalLengthCm'] *30,
      alpha = 0.4,
      label = species
  )

plt.title("Iris Flower Features")

plt.legend(
    title = "Species"
)
plt.savefig("Bubble_chart.png", dpi=300, bbox_inches='tight')
plt.plot()
```

### 3. Heat Map

- Create a correlation heatmap for all numeric features.

- Title: "Heatmap of Iris features"



```python
# Heat Map
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

numeric_data = data.select_dtypes(["number"])

corr_matrix = numeric_data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,        # Show correlation values
    cmap="coolwarm",   # Color map for better contrast
    linewidths=0.5
)

plt.title("Heatmap of Iris Features")
plt.savefig("heat_map.png", dpi=300, bbox_inches='tight')
plt.show()
```

### 4. Box Plot

- Create box plots for petal length by species.

- Title: "Petal Length Distribution by Species"

- Create box plots showing the distribution of petal length for each species.

- Clearly represent the median, first quartile (Q1), third quartile (Q3), and the interquartile range (IQR).

- Display outliers using distinct marker symbols for easy identification.

- Use different colors for each species to ensure clear visual distinction.

```python
# 4. Box Plot
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

species_list = data["Species"].unique()

petal_length_data= []

for species in species_list:
  petal_length_data.append(
      data[data["Species"] == species]["PetalLengthCm"]
  )

plt.boxplot(
    petal_length_data,
    labels=species_list,
    patch_artist=True,
    showfliers=True,
    flierprops=dict(          
        marker='o',
        markerfacecolor='red',
        markersize=6,
        linestyle='--'
    )
)


plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal length (cm)")

plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.savefig("box_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```

### 5. Violin Plot

- Create violin plots for petal width across species.

- Title: "Petal Width Distribution by Species"

```python
#violin plot 
from matplotlib.legend import Legend
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")

sns.violinplot(
    x="Species",
    y="PetalWidthCm",
    data=data,
    hue="Species",
    palette="viridis",
    legend=False

)


plt.title("Petal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Width (cm)")

plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.savefig("violin_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```

### 6. Pair Plot

- Create pairwise scatter plots between all numeric features.

- Create pairwise scatter plots between all numeric features of the Iris dataset: Sepal Length, Sepal Width, Petal Length and Petal Width.

- Display feature-to-feature relationships across all possible combinations.

- Use a consistent axis scale for all subplots to ensure fair visual comparison. Apply uniform marker style and size for visual consistency.

- Use different colors for different species to enable class-wise visual separation.

- Maintain color consistency for each species across all subplots.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab4/Datasets/Iris.csv")


sns.pairplot(
    data,
    vars=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"],
    hue="Species",          
    palette="Set1",        
    markers="o",            
    plot_kws={
        "s": 40,            
        "alpha": 1,
        # "edgecolor": "black"      
    },
    diag_kind="kde"    
)   

plt.suptitle("Pairwise Feature Relationships in Iris Dataset",y=1.02)
plt.savefig("pair_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```
