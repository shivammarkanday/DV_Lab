### 1. Write code for line chart to generate 70 numbers with X- axis and Y -axis is mx+c, were m=2 and c=0.3, Draw the four different line charts using matplotlib library. (Take Y1=X, Y2=X**2, Y3=X**3 and Y4=sqrt(X)).

```python
import numpy as np 
import matplotlib.pyplot as plt 

x = np.arange(1,71)

y1 = x
y2 = x**2
y3=x**3
y4= np.sqrt(x)

plt.plot(x, y1, label="Y = X")
plt.plot(x, y2, label="Y = X²")
plt.plot(x, y3, label="Y = X³")
plt.plot(x, y4, label="Y = √X")

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Line Charts for Different Mathematical Functions")

plt.legend()
plt.grid(visible="True", alpha=0.6)

plt.tight_layout()

plt.savefig("Question_1.png")
plt.show()
```

### 2. Plot the function y = x² - 4x + 3 for x ranging from -2 to 6. Identify and annotate the minimum point of the graph. Add appropriate labels, title, and grid.

```python
import numpy as np 
import matplotlib.pyplot as plt 

x = np.linspace(-2,6,100)

y = x**2 - 4*x + 3 

plt.plot(x,y,label="x**2 - 4*x + 3")

x_min = 2
y_min = x_min**2 - 4*x_min + 3

plt.scatter(x_min,y_min,color="red")
plt.annotate(
    f"min point ({x_min},{y_min})",
    xy=(x_min,y_min),
    xytext=(3, y_min + 5),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = x² - 4x + 3")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.savefig("Question_2.png")
plt.show()
```

### 3. Plot sin(x) and cos(x) from 0 to 2π on the same graph. Use fill_between() to shade the area between the two curves where sin(x) is greater than cos(x). Add labels and a title.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 400)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y_sin, label="sin(x)", color="blue")
plt.plot(x, y_cos, label="cos(x)", color="orange")

plt.fill_between(
    x,
    y_sin,
    y_cos,
    where=(y_sin > y_cos),
    interpolate=True,
    alpha=0.3,
    color="green"
)

plt.xlabel("x")
plt.ylabel("Value")
plt.title("sin(x) and cos(x) with Shaded Region where sin(x) > cos(x)")
plt.legend()
plt.tight_layout()
plt.grid(True)


plt.savefig("Question_3.png")
plt.show()
```

### 4. Consider monthly temperature data for the year 2023 and plot it using a time-series line graph.
### months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
### temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]
### Label the x-axis as Months and y-axis as Temperature (°C). Format the x-axis to show month names instead of numbers.

```python
import numpy as np
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]

plt.plot(months, temperature, marker='o', color='steelblue')

plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.title("Monthly Temperature Variation in 2023")


plt.grid(True,alpha=0.3)
plt.tight_layout()


plt.savefig("Question_4.png")
plt.show()
```

### 5. Plot three mathematical functions on the same graph:
### - y = x² (Red, Dashed Line)
### - y = 2x + 1 (Blue, Solid Line)
### -  y = sqrt(x) (Green, Dotted Line)
### Add a legend, labels, and title.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1,100,100)

y1 = x**2
y2 = 2*x+1
y3 = np.sqrt(x)

plt.plot(x,y1,color="red",linestyle="dashed")
plt.plot(x,y2,color="blue",linestyle="solid")
plt.plot(x,y3,color="green",linestyle="dotted")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Multiple y graph ")
plt.legend()
plt.grid(True)
plt.tight_layout()


plt.savefig("Question_5.png")
plt.show()
```

### 6. Create a dual-axis plot where:
### - The primary y-axis represents the monthly rainfall (in mm) in 2023 as a bar chart.
### - The secondary y-axis represents the average temperature (in °C) as a line graph on the same plot.
### months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
### rainfall = [78, 62, 55, 41, 32, 140, 300, 280, 150, 90, 85, 80]
### temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]
### Label both y-axes and use different colors for bars and the line.

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun','Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
rainfall = [78, 62, 55, 41, 32, 140, 300, 280, 150, 90, 85, 80]
temperature = [12, 14, 18, 24, 30, 34, 32, 31, 28, 22, 16, 13]

fig,ax1 = plt.subplots()

ax1.bar(months,rainfall,color="steelblue",label="Rainfall(mm)")
ax1.set_xlabel("Months")
ax1.set_ylabel("Rainfall (mm)", color="steelblue")
ax1.tick_params(axis='y',labelcolor="steelblue")

ax2 = ax1.twinx()

ax2.plot(months, temperature, color="firebrick", marker="o", label="Temperature (°C)")
ax2.set_ylabel("Temperature (°C)", color="firebrick")
ax2.tick_params(axis='y',labelcolor="firebrick")

plt.title("Monthly Rainfall and Temperature in 2023")
plt.tight_layout()


plt.savefig("Question_6.png")
plt.show()
```

### 7: Draw a house using matplotlib line charts:

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))

# Walls (rectangle)
plt.plot([1, 5], [1, 1], color="brown", label="Walls")   
plt.plot([1, 1], [1, 4], color="brown")                 
plt.plot([5, 5], [1, 4], color="brown")                
plt.plot([1, 5], [4, 4], color="brown")                

# Roof (triangle)
plt.plot([1, 3], [4, 6], color="darkred", label="Roof")
plt.plot([3, 5], [6, 4], color="darkred")

# Door (rectangle)
plt.plot([2, 3], [1, 1], color="blue", label="Door")
plt.plot([2, 2], [1, 3], color="blue")
plt.plot([3, 3], [1, 3], color="blue")
plt.plot([2, 3], [3, 3], color="blue")

# Window 1
plt.plot([4, 4.5], [3, 3], color="teal", label="Window 1")
plt.plot([4, 4], [3, 3.5], color="teal")
plt.plot([4.5, 4.5], [3, 3.5], color="teal")
plt.plot([4, 4.5], [3.5, 3.5], color="teal")

# Window 2
plt.plot([4, 4.5], [1.5, 1.5], color="purple", label="Window 2")
plt.plot([4, 4], [1.5, 2], color="purple")
plt.plot([4.5, 4.5], [1.5, 2], color="purple")
plt.plot([4, 4.5], [2, 2], color="purple")

plt.title("House Using Line Plots with Multiple Colors")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(loc="upper right")
plt.grid(True)
plt.tight_layout()


plt.savefig("Question_7.png")
plt.show()
```
