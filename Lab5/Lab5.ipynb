## **Dataset 1: Penguins Dataset**





```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("penguins")

df.head()
```

### 1. Catplot

- Create a bar catplot to compare average body mass across species and sex.

- Title: "Average Body Mass by Species and Sex"

- X-axis: Species

- Y-axis: Body Mass (g)

- Hue: Sex

```python
g = sns.catplot(
    data = df,
    x = "species",
    y = "body_mass_g",
    kind="bar",
    hue="sex",
    palette="Set2",
)

g.fig.suptitle("Average Body Mass by Species and Sex",y=1.02)
g.set_axis_labels("Species", "Body Mass (g)")

plt.show()
```

### 2. Stripplot

- Visualize flipper length distribution by species and sex.

- Title: "Flipper Length Distribution by Species and Sex"

- X-axis: Species

- Y-axis: Flipper Length (mm)

- Hue: Sex

```python
plt.figure(figsize=(8,6))

sns.stripplot(
    data = df,
    x = "species",
    y = "flipper_length_mm",
    hue = "sex",
    dodge=True,
    jitter=True,
    palette="Set1",

)

plt.title("Flipper Length Distribution by Species and Sex")
plt.xlabel("Species")
plt.ylabel("Flipper Length (mm)")

plt.legend(title="sex")

plt.show()
```

### 3. Heatmap

- Display correlation matrix of numerical features (bill length, bill depth, flipper length, mass)

- Annotate values, use colorbar to represent correlation strength.

- Title: "Correlation Heatmap of Penguin Features"

```python
numeric_data = df[[
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
    ]]

corr_matrix= numeric_data.corr(
)

sns.heatmap(
    data = corr_matrix,
    annot = True,
    cbar = True,
    cmap = "binary",
    linewidths=0.1
)

plt.title("Correlation Heatmap of Penguin Features")
plt.show()
```

### 4. Clustermap

- Create a clustered heatmap to group features based on similarity.

- Title: "Clustered Correlation Map of Penguin Features"

```python
numeric_data = df[[
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]]

corr_matrix = numeric_data.corr()

g = sns.clustermap(
    corr_matrix,
    annot=True,
    cmap="binary",
    linewidths=0.5,
    figsize=(8, 8)
)

g.fig.suptitle("Clustered Correlation Map of Penguin Features", y=1.02)
plt.show()
```

## **Dataset 2: Tips Dataset**






```python
df_tips = sns.load_dataset("tips")
df_tips.head()
```

### 5. Box Plot

- Compare total bill distribution across days.

- Title: "Total Bill Distribution by Day"

- X-axis: Day

- Y -axis: Total Bill

```python
sns.boxplot(data = df_tips, x="day", y="total_bill")
plt.title("Total Bill Distribution by Day")
plt.show()
```

### 6. Violin Plot

- Show tip distribution by smoker status.

- Title: "Tip Distribution by Smoker Status"

- X-axis: Smoker

- Y-axis: Tip

```python
plt.figure(figsize=(8, 6))

sns.violinplot(
    data=df_tips,
    x="smoker",
    y="tip",
    hue="smoker",
    palette="Set2"
)

plt.title("Tip Distribution by Smoker Status")
plt.xlabel("Smoker")
plt.ylabel("Tip")
plt.tight_layout()

plt.show()
```

### 7. KDE Plot

- Plot density of total bill amounts split by gender.

- Title: "Density of Total Bill by Gender"

- X-axis: Total Bill

- Hue: Gender

```python
plt.figure(figsize=(8, 6))

sns.kdeplot(
    data=df_tips,
    x="total_bill",
    hue="sex",
    fill=True,
    common_norm=False,
    alpha=0.5,
    palette="Set2"
)

plt.title("Density of Total Bill by Gender")
plt.xlabel("Total Bill")
plt.ylabel("Density")

plt.show()
```

### 8. Joint Plot

- Explore relationship between total bill and tip.

- Title: "Total Bill vs Tip Relationship"

- X-axis: Total Bill

- Y-axis: Tip

- Hue: Gender

```python
g = sns.jointplot(
    data=df_tips,
    x="total_bill",
    y="tip",
    hue="sex",
    kind="scatter",
    palette="Set2",
    height=6
)

g.fig.suptitle("Total Bill vs Tip Relationship", y=1.02)

plt.show()
```

## **Dataset 3: IRIS Dataset**


### 9. Pairplot

- Create pairwise relationships among all numeric features.

- Title: "Pairwise Relationships in Iris Dataset"

- Hue: Species

```python
iris = sns.load_dataset("iris")

g = sns.pairplot(
    data=iris,
    hue="species",
    palette="Set1",
    diag_kind="hist"
)

g.fig.suptitle("Pairwise Relationships in Iris Dataset", y=1.02)

plt.show()
```

### 10. Swarm Plot

- Visualize petal length distribution across species.

- Title: "Petal Length Distribution by Species"

- X-axis: Species

- Y -axis: Petal Length

```python
plt.figure(figsize=(8, 6))

sns.swarmplot(
    data=iris,
    x="species",
    y="petal_length",
    palette="Set1"
)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()
```

### 11. Bubble Chart (Scatterplot with size)

- Sepal length vs sepal width

- bubble size = petal length

- hue = species.

- Title: "Sepal vs Petal Dimensions in Iris"

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="sepal_width",
    size="petal_length",
    hue="species",
    sizes=(20, 200),
    alpha=1,
    palette="Set1"
)

plt.title("Sepal vs Petal Dimensions in Iris")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.legend(title="Species")

plt.show()

```
