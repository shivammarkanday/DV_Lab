### 1. Line Plot

- Plot the rent (“Rent” column) against the size (“Size” column).

- Use different line styles (solid, dashed, dotted) for better visualization.

- Add gridlines to the plot.

- Title: "Rent vs. Size"

- Label the x-axis as "Size (sq ft)" and the y-axis as "Rent (INR)"

- Set the legend position to the top-right corner.

```python
import pandas as pd 
import matplotlib.pyplot as plt 

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

sorted_data = data.sort_values(by='Size')

sample_data = sorted_data.head(50)

size = sample_data["Size"]
rent = sample_data["Rent"]

plt.plot(size, rent, linestyle='solid', label='Solid line')

plt.title("Rent vs Size")
plt.xlabel("Size (sq ft)")
plt.ylabel("Rent (INR)")

plt.grid(True)

plt.legend()

plt.savefig("line_plot_rent_vs_size.png")

plt.show()
```

### 2. Bar Chart

- Create a bar chart to display the average rent for each furnishing status
(“Furnishing Status”).

- Use different colors for each bar.

- Title: "Average Rent by Furnishing Status"

- Label the x-axis as "Furnishing Status" and the y-axis as "Average Rent (INR)"

```python
import pandas as pd 
import matplotlib.pyplot as plt 

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

avg_rent = data.groupby("Furnishing Status")["Rent"].mean()

plt.bar(
    avg_rent.index,
    avg_rent.values,
    color=["skyblue","lightgreen","salmon"]
)

plt.title("Average Rent by Furnishing Status")
plt.xlabel("Furnishing Status")
plt.ylabel("Average Rent (INR)")

plt.tight_layout()
plt.savefig("Bar_chart.png")
plt.show()
```

### 3. Scatter Plot

- Plot the number of bathrooms (“Bathroom” column) against the rent (“Rent”
column).

- Use different marker styles for points representing flats in different cities (“City”
column).

- Title: "Bathrooms vs. Rent"

- Label the x-axis as "Number of Bathrooms" and the y-axis as "Rent (INR)"

- Use a legend to indicate cities and place it below the plot.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

cities = data["City"].unique()

markers = ["o", "s", "^", "D", "v", "P", "X", "*"]

for city, marker in zip(cities, markers):
    subset = data[data["City"] == city]
    plt.scatter(
        subset["Bathroom"],          
        subset["Rent"],              
        marker=marker,              
        alpha=0.7,
        label=city
    )

plt.title("Bathrooms vs. Rent")
plt.xlabel("Number of Bathrooms")
plt.ylabel("Rent (INR)")

plt.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.15),
    ncol=3,
    title="City"
)

plt.tight_layout()
plt.savefig("scatter_plot.png")
plt.show()
```

### 5. Stacked Bar Chart

- Create a stacked bar chart to display the count of flats by furnishing status for
each city.

- Use different colors for each furnishing status.

- Title: "Flat Count by Furnishing Status and City"

- Label the x-axis as "City" and the y-axis as "Flat Count"


```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

pivot_table = pd.pivot_table(
    data,
    index = "City",
    columns="Furnishing Status",
    values="Rent",
    aggfunc= "count",
    fill_value=0
)

pivot_table.plot(
    kind="bar",
    stacked="True",
    color=["dodgerblue", "forestgreen", "crimson"]
)

plt.title("Flat Count by Furnishing Status and City")
plt.xlabel("City")
plt.ylabel("Flat Count")

plt.tight_layout()
plt.savefig("stacked_bar_chart.png")
plt.plot()
```

### 6. Pie Chart

- Create a pie chart to show the proportion of flats in different cities (“City” column).

- Use a different color for each segment.

- Add percentage labels to each segment.

- Title: "Distribution of Flats by City"

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

city_count = data["City"].value_counts()

plt.pie(
    city_count.values,
    labels=city_count.index,
    autopct="%1.1f%%",
    colors=plt.cm.tab20.colors,
    startangle=90,
)

plt.title("Distribution of Flats by City")
plt.tight_layout()
plt.savefig("pie_chart.png")
plt.show()
```

### 7. Stacked Plot

- Create a stacked plot to visualize rent, size, and number of bathrooms for the first 10 rows of the dataset.

- Title: "Stacked Plot of Rent, Size, and Bathrooms"

- Label the x-axis as "Index" and the y-axis as "Values"

- Use different colors for each variable.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

data_head = data.head()

x_axis = data_head.index

rent = data_head["Rent"]
size = data_head["Size"]
bathrooms = data_head["Bathroom"]

plt.stackplot(
    x_axis,
    rent,
    size,
    bathrooms,  
    labels=["Rent","Size","Bathroom"],
    colors=["steelblue", "seagreen", "firebrick"]
)

plt.title("Stacked Plot of Rent, Size, and Bathrooms")
plt.xlabel("Index")
plt.ylabel("Values")

plt.legend(
    loc="upper left"
)
plt.savefig("stacked_plot.png")
plt.show()
```
