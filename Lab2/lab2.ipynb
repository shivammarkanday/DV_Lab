### 1. Line Plot

- Plot the rent (“Rent” column) against the size (“Size” column).

- Use different line styles (solid, dashed, dotted) for better visualization.

- Add gridlines to the plot.

- Title: "Rent vs. Size"

- Label the x-axis as "Size (sq ft)" and the y-axis as "Rent (INR)"

- Set the legend position to the top-right corner.

```python
import pandas as pd 
import matplotlib.pyplot as plt 

data = pd.read_csv("/Users/shivammarkanday/Library/CloudStorage/OneDrive-AdaniUniversity/OneDrive College/SEM 6 (2026)/DV/Jupyter Notebook/Lab2/Datasets/House_Rent_Dataset.csv")

sorted_data = data.sort_values(by='Size')

sample_data = sorted_data.head(50)

size = sample_data["Size"]
rent = sample_data["Rent"]

# plt.plot(size, rent, linestyle='solid', label='Solid line')
# plt.plot(size, rent, linestyle='dashed', label='Dashed Line')
plt.plot(size, rent, linestyle='dotted', label='Dotted Line')

plt.title("Rent vs Size")
plt.xlabel("Size (sq ft)")
plt.ylabel("Rent (INR)")

plt.grid(True)

plt.legend()

plt.savefig("line_plot_rent_vs_size.png")

plt.show()

```
